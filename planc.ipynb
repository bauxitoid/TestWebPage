```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider
import ipywidgets as widgets
from IPython.display import display

# Fyzikální konstanty
h = 6.626e-34  # Planckova konstanta [J·s]
c = 3e8        # rychlost světla [m/s]
k = 1.381e-23  # Boltzmannova konstanta [J/K]
wien_const = 2.897e-3  # Wienova konstanta [m·K]

# Bezpečný výpočet Planckova zákona
def planck(wavelength, T):
    a = 2.0 * h * c**2
    b = h * c / (wavelength * k * T)
    with np.errstate(over='ignore', divide='ignore', invalid='ignore'):
        expo = np.expm1(b)
        expo = np.where(expo == 0, np.inf, expo)
        intensity = a / (wavelength**5 * expo)
        intensity = np.nan_to_num(intensity, nan=0.0, posinf=0.0, neginf=0.0)
    return intensity

# Funkce pro vykreslení obou spekter
def plot_spectra(T1, T2):
    wavelengths = np.linspace(1e-9, 3e-6, 500)  # od 1 nm do 3 µm
    intensity1 = planck(wavelengths, T1)
    intensity2 = planck(wavelengths, T2)

    lambda_max1 = wien_const / T1
    lambda_max2 = wien_const / T2

    plt.figure(figsize=(9, 5))
    plt.plot(wavelengths * 1e9, intensity1, label=f"T = {T1} K", color='blue')
    plt.axvline(lambda_max1 * 1e9, color='blue', linestyle='--', label=f"λₘₐₓ ({T1} K) ≈ {lambda_max1*1e9:.0f} nm")

    plt.plot(wavelengths * 1e9, intensity2, label=f"T = {T2} K", color='red')
    plt.axvline(lambda_max2 * 1e9, color='red', linestyle='--', label=f"λₘₐₓ ({T2} K) ≈ {lambda_max2*1e9:.0f} nm")

    plt.title('Porovnání Planckova spektra pro dvě teploty')
    plt.xlabel('Vlnová délka [nm]')
    plt.ylabel('Intenzita [W/m³]')
    plt.grid(True)
    plt.legend()
    plt.tight_layout()
    plt.show()

# Dva posuvníky pro teploty
slider1 = FloatSlider(value=3000, min=1000, max=10000, step=100, description='Teplota T₁ [K]')
slider2 = FloatSlider(value=6000, min=1000, max=10000, step=100, description='Teplota T₂ [K]')
ui = widgets.HBox([slider1, slider2])
out = widgets.interactive_output(plot_spectra, {'T1': slider1, 'T2': slider2})

display(ui, out)
```
